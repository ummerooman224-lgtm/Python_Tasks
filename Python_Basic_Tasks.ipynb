```python
# ==========================================
# TASK 1: CINEMA TICKET PRICE
# ==========================================

name = input("Enter your name: ")
age = int(input("Enter your age: "))

if age < 5 or age > 100:
    ticket_price = "Free"
elif age >= 5 and age <= 12:
    ticket_price = "Rs. 300"
elif age >= 13 and age <= 59:
    ticket_price = "Rs. 700"
elif age >= 60 and age <= 100:
    ticket_price = "Rs. 400"

print()
print("Customer:", name)
print("Ticket Price:", ticket_price)


# ==========================================
# TASK 2: TEMPERATURE CONVERTER
# ==========================================

celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32

if celsius < 10:
    weather = "Very Cold"
elif celsius >= 10 and celsius <= 19:
    weather = "Cold"
elif celsius >= 20 and celsius <= 29:
    weather = "Normal"
else:
    weather = "Warm / Hot"

print()
print("Temperature in Fahrenheit:", fahrenheit)
print("Weather Condition:", weather)


# ==========================================
# TASK 3: MARKS & GRADE CALCULATOR
# ==========================================

print("\n" + "="*30 + "\n")

marks = float(input("Enter your percentage/marks: "))

if marks >= 80:
    grade = "A+"
elif marks >= 70:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "Fail"

print()
print("Your Grade is:", grade)


# ==========================================
# TASK 4: EVEN OR ODD CHECKER
# =========================================
num = int(input("Enter any number: "))

if num % 2 == 0:
    result = "Even Number"
else:
    result = "Odd Number"

print()
print("Number Type:", result)


# ==========================================
# TASK 5: SIMPLE CALCULATOR
# ==========================================



num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")

if operation == "+":
    ans = num1 + num2
elif operation == "-":
    ans = num1 - num2
elif operation == "*":
    ans = num1 * num2
elif operation == "/":
    if num2 != 0:
        ans = num1 / num2
    else:
        ans = "Cannot divide by zero"
else:
    ans = "Invalid operation"

print()
print("Result:", ans)
```
